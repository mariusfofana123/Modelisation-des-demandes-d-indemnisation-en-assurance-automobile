```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler

# 1. Chargement des données
insurance = pd.read_csv("car_insurance.csv")  # Ajuste le nom du fichier si besoin

# 2. Imputation des valeurs manquantes par la moyenne
insurance["credit_score"] = insurance["credit_score"].fillna(insurance["credit_score"].mean())
insurance["annual_mileage"] = insurance["annual_mileage"].fillna(insurance["annual_mileage"].mean())
```

```python
# Étape 2 : Validation DataCamp (Corrigée)
features = [col for col in insurance.columns if col not in ["id", "outcome"]]

best_feature = None
best_accuracy = 0.0

for feature in features:
    X_single = insurance[[feature]].copy()
    
    # Si la colonne n'est pas déjà numérique, conversion en codes de catégories
    if not pd.api.types.is_numeric_dtype(X_single[feature]):
        X_single[feature] = X_single[feature].astype('category').cat.codes
        
    y = insurance["outcome"]
    
    # Entraînement univarié
    model = LogisticRegression(max_iter=1000)
    model.fit(X_single, y)
    
    # Évaluation de l'accuracy
    preds = model.predict(X_single)
    acc = accuracy_score(y, preds)
    
    # Conservation du meilleur score
    if acc > best_accuracy:
        best_accuracy = acc
        best_feature = feature

# Création du DataFrame demandé par DataCamp
best_feature_df = pd.DataFrame({
    "best_feature": [best_feature],
    "best_accuracy": [best_accuracy]
})

print("--- RÉSULTAT DATACAMP ---")
print(best_feature_df)
```
